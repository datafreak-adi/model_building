## Building my own multiple regression 

```python
class Meramulti:

    def __init__(self):

        self.intercept_= None
        self.slope_= None

    def fit(self, X_train, Y_train):

        np.insert(X_train,0,1,axis=1)

        b_value= np.linalg.inv(np.dot(X_train.T, X_train)).dot(X_train.T).dot(Y_train)

        self.intercept_= b_value[0]
        self.slope_= b_value[1: ]

        print(f"The regression line thus obtaines has a intercept of {self.intercept_} and slopes of {self.slope_}")
        

    def predict(self,X_test):

        y_pred= np.dot(X_test,self.slope_) + self.intercept_
        return y_pred
```

```python
lrg= Meramulti()
```

```python
lrg.fit(X_train, Y_train)
```

```python
lrg.predict(X_test.iloc[0:4].values)
```
