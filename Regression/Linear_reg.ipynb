```python
class MeraLr():

    def __init__ (self):
        self.m= None
        self.b= None
        
    def fit(self, X_train, Y_train):

        X_mean= X_train.mean()
        Y_mean= Y_train.mean()
        num=0
        den=0
        # Calculating slope
    
        for i in range(X_train.shape[0]):

          
           
            num = num + np.sum((X_train[i]- X_mean) *( Y_train[i] - Y_mean))
            den= den + np.sum((X_train[i]-X_mean)**2)

        #slope
        self.m= num/den
        
        # calculating intercept
        self.b= Y_mean - self.m* X_mean
        print( f" The Regression line for the given dataset accoring to our model has a slope of {self.m} and intercept {self.b}") 


    def predict(self,X_test):

        return self.m* X_test +self.b

        
```

```python

lr= MeraLr()
```
